```python
# np unique
np.unique(arr)

values, counts = np.unique(arr, return_counts= True)
print(values, counts)
```

```python
# np.expand_dims() add a new dimension
arr3 = np.array([1,2,3])
arr3.shape

# convert it to a row vector
arr3 = np.expand_dims(arr3, axis=0)
arr3.shape

# convert into a col vector
arr3 = np.array([1,2,3])
np.expand_dims(arr3, axis=1)
```

```python
# np.cumsum cumulative sum

np.cumsum(arr)
```

```python
# cumprod
np.cumprod(arr)
```

```python
# np.percentile() - Finds percentile values
np.percentile(arr, 50)
np.percentile(arr,25)
np.percentile(arr,75)
```

```python
# np.histogram
hist, bins = np.histogram(arr, bins=5)
print(hist)
print(bins)

# Histogram counts tell you how many values fall into each interval.

# Very useful before plotting histograms.
```

```python
# np.corrcoef()
# find correlation between variable
np.corrcoef(arr, arr)

# Values range from -1 to 1:

# 1 → perfect positive correlation
# 0 → no linear relationship
# -1 → perfect negative correlation
```

```python
# np.isin()
# cheks membership
np.isin(arr, [2,5])
```

```python
# np.flip
print(arr)
print(np.flip(arr))
```

```python
# np.ptp
# peak to peak
# max - min
np.ptp(arr)
```

```python
# set functions
# np.union1d()
np.union1d(arr, arr)
# np.intersect1d()
np.intersect1d(arr, arr2)
# np.setdiff1d()
np.setdiff1d(arr, arr2)
# np.setxor1d()
# Values that are unique to each array (not common).
np.setxor1d(arr, arr2)
```
