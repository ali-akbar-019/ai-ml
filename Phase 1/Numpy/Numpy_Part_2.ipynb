```python
# lets make something fun
x = np.linspace(-1.0, 1.0, 1000)
y_top = (x**2)**(1/3) + np.sqrt(1 - x**2)
y_bottom = (x**2)**(1/3) - np.sqrt(1 - x**2)

plt.plot(x, y_top, color="red")
plt.plot(x, y_bottom, color="red")
plt.title("Python Heart Graph")
plt.grid(True)
plt.axis("equal")
plt.show()
```

```python
# the infinity sign
t = np.linspace(0, 2*np.pi, 1000)
x = 2 * np.cos(t) / (1 + np.sin(t)**2)
y = 2 * np.sin(t) * np.cos(t) / (1 + np.sin(t)**2)

plt.plot(x, y, color="blue")
plt.title("Infinite Sign")
plt.grid(True)
plt.axis("equal")
plt.show()
```

```python
# problem 1 - student marks analysis
students_marks = np.random.randint(40, 100, (5,6))
print("All Student Marks\n",students_marks)

#
print("Shape:",students_marks.shape)
print("Average of the Marks:",students_marks.mean())
print("Highest Marks:", students_marks.max())
print("Lowest Marks:", students_marks.min())
print("Marks Greater than 50 ", np.where(students_marks < 50, 50, students_marks))

print("Sort Marks: ", np.sort(students_marks))

# graph
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = np.mean(students_marks, axis=1)

# Bar chart for average marks
plt.bar(x, y, color='skyblue', edgecolor='black')
plt.xlabel("Student Number")
plt.ylabel("Average Marks")
plt.title("Average Marks of Students")
plt.xticks(x)  # Show all student numbers
plt.grid(axis='y', linestyle='--', alpha=0.7)
```
