```python
# reshape
arr = np.arange(12)
print(arr)
print(arr.shape)
arr = arr.reshape(3,4)
print(arr, arr.shape)
# but ye 12 divisible ha isi leye, 3 X 4 = 12
```

```python
# array operations
arr1 = np.array([1,2,3,4,5,6])
print(arr1)
arr1 = arr1 + 5
print(arr1)

# subtraction
arr1 = arr1 - 10
print(arr1)

# multiplication
arr1 = arr1 * 2
print(arr1)

# division
arr1 = arr1 / 10
print(arr1)

# power
arr1 = arr1 ** 2
print(arr1)

# modulus
arr1 = arr1 % 2
print(arr1)
```

```python
# comparision operations
arr = np.arange(0,15)
arr > 5

arr < 10

arr == 8 #jon sa el ho ga wo true dikae ga array me

arr != 5

arr >= 5
```

```python
mask1 = arr > 5
print(arr[mask1])
```

```python
# vector operations
a = np.arange(1,10)
b = np.arange(10,19)
print(a + b)
print(a - b)
print(a * b)
print(a / b)
```

```python
# mathematical functions
arr = np.array([2,5,7,10])
print(arr.min())
print(arr.max())
print(arr.sum())
print(arr.prod())
```

```python
# statistical functions
arr.mean() #average
# median middle value after sorting
np.median(arr)

# standard deviation
arr.std()
# variance
arr.var()
```

```python
# trignometric functions
angles = np.array([0, np.pi/2, np.pi])
np.sin(angles)
np.cos(angles)
np.tan(angles)
np.deg2rad(90)
```

```python
# dot product

np.dot(a,b)
```

```python
# logarithms & exponentials
np.log(arr)
print(np.log10(arr))
print(np.exp(arr)) #e ^ x
```

```python
# rounding function
np.round(arr)
print(np.floor(arr))
print(np.ceil(arr))
```

```python
# indexing
arr = np.array([10,20,30,40,50])
print(arr[0])
print(arr[2])
print(arr[4])
print(arr[-1])
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr[1,2])
```

```python
# slicing
arr[1:4]
arr[0, 0::2]
```

```python
# iterating
for num in arr:
  print(num)
```

```python
for value in np.nditer(arr):
  print(value)
  # np.nditer() is useful for iterating through arrays of any dimension.
```

```python
# stacking
a = np.arange(1,10)
b = np.arange(10, 19)

# horizontal stacking
hs =np.hstack((a,b))
print(hs)
vs = np.vstack((a,b))
print(vs)
```

```python
# splitting
arr = np.array([1,2,3,4,5,6])
np.hsplit(arr,3)
matrix = np.array([
    [1,2],
    [3,4],
    [5,6],
    [7,8]
])

np.vsplit(matrix,2)
```
