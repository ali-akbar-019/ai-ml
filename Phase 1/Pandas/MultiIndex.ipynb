```python
import pandas as pd
import numpy as np
```

```python
# multiindexing allows pandas to use multiple levels of indexes instead of one

# it lets us represent higher dimensional/hierarchical data using lower dimensional
# pandas structures
```

```python
# create a multiindex using from_tuples()

index = pd.MultiIndex.from_tuples([
    ("Islamabad", 2024),
    ("Islamabad", 2025),
    ("Lahore", 2024),
    ("Lahore", 2025)
])

# now create a series
sales=pd.Series(
    [100,150,200,250],
    index=index
)
sales
# sales.loc["Islamabad"]
```

```python
# understanding the levels
print(sales.index)
print("\n"*2)
print(sales.index.names)
print("\n"*2)

# you can give names
sales.index.names = ["City", "Year"]
print(sales.index.names) #now the indexes has the names

print("\n"*3)
print(sales)
```

```python
# level 0 -> city
# level 1 -> year
```

```python
# accessing multiindex Data
# this is where multiindex becomes useful
# get everything from isb
sales.loc["Islamabad"]
```

```python
# get isb 2024
sales.loc["Islamabad", 2024]
```

```python
# Partial Selection
# you dont have ot provide entry level
sales.loc["Lahore"]
```

```python
# multiindex from product
# cartisian product kar deta ha product
pd.MultiIndex.from_product([
    ["Islamabad", "Lahore"],
    [2024,2025,2026]
])
```

```python
# multi index data frame
# multi index is not only for series
index = pd.MultiIndex.from_product([
    ["Islamabad", "Lahore"],
    [2024, 2025]
],
names=["City", "Year"])
df = pd.DataFrame(
    {
        "Sales": [100,150,200,250],
        "Profit": [20,30,40,50]
    },
    index=index
)
df
```

```python
# selecting from multiindex dataframe
df.loc["Islamabad"]


df.loc[("Islamabad", 2025)]
```

```python
df["Sales"]
```

```python
df.loc["Islamabad", "Sales"]
```

```python
# why not just keep city and year as normal columns
# You absolutely can.

# For example:

# City        Year       Sales
# Islamabad   2024       100
# Islamabad   2025       150
# Lahore      2024       200
# Lahore      2025       250

# That's often easier for beginners.

# MultiIndex becomes useful when your data naturally has hierarchical
#  relationships and you repeatedly need to select/aggregate by those levels.
```

```python
# unstack
#index level column ban jata ha - most inner wala col ban jaye ga
print(df)
print("\n"*3)
print(df.unstack())
# ab yaha pe jo index wala yer tha wo ja k column ban gaya ha
print("\n"*3)
df2 = df.unstack()
df2.loc["Islamabad", "Sales"]
```

```python
# unstack with a specific level
sales
s2 = sales.unstack(level="Year")
# the year become the columns
s2.loc["Islamabad"]
s2.loc["Islamabad", 2024]
```

```python
# stack - opposite of the unstack
# in it the column become the index
print("DF\n",df,"\n\n")
print("df.stack()\n",df.stack(),"\n\n")
```

```python
# unstack → index → columns

# stack → columns → index
#
# Stack = stack columns down into the index. -- wider

# Unstack = pull an index level out into columns. -- longer
```

```python
# Why do we care about Long vs Wide?

# Different operations prefer different formats.

# Long format

# Excellent for:

# analysis
# grouping
# plotting
# statistical work
# Wide format

# Often easier for:

# reports
# presentations
# human reading

# You'll constantly convert between them.
```

```python
# melt convert wide to long
df2 = pd.DataFrame({
    "Name": ["Ali", "Zarak"],
    "Maths": [90, 100],
    "English":[100,100],
    "Science": [56,100]
})
df2 #this is a wide data
# now lets use the melt on it to  convert it to long
long_df2 = df2.melt(id_vars="Name")
long_df2
```

```python
# give a better column name
long_df2 = df2.melt(
    id_vars="Name",
    var_name="Subject",
    value_name="Marks"
)
long_df2
```

```python
# melt with mult columns
df = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Gender": ["Male", "Female"],
    "Math": [90,80],
    "English": [85,95]
})
df.melt(
    id_vars=["Name", "Gender"],
    var_name="Subject",
    value_name="Marks"
)
```

```python
# pivot table
# a pivot table summarizes data across multiple categories
df = pd.read_csv("train.csv")
```

```python
# average fare for each passenger class and gender
pd.pivot_table(
    df,
    values="Fare",
    index="Pclass",
    columns="Sex",
    aggfunc="mean"
)
```

```python
pd.pivot_table(
    df,
    values="Fare",
    index="Pclass",
    columns="Sex",
    aggfunc="count"
)
```

```python
# pivot table with multiple aggregations
pd.pivot_table(
    df,
    values="Fare",
    index="Pclass",
    columns="Sex",
    aggfunc=["mean", "max"]
)
```

```python
# pivot table with multiple indexes
pd.pivot_table(
    df,
    values="Fare",
    index=["Pclass","Embarked"],
    columns="Sex",
    aggfunc="mean"
)
```

```python
# fill na
# pivot table can contain NaN
pd.pivot_table(
    df,
    values="Fare",
    index="Pclass",
    columns="Sex",
    aggfunc="mean",
    fill_value=0
)
```

```python
# Pivot Table vs GroupBy

# This is another common confusion.

# GroupBy

# Best when you want to perform grouped calculations.

# df.groupby("Pclass")["Fare"].mean()
# Pivot Table

# Best when you want a table/matrix showing relationships between categories.

# pd.pivot_table(
#     df,
#     values="Fare",
#     index="Pclass",
#     columns="Sex",
#     aggfunc="mean"
# )

# Think:

# groupby
# → analysis

# pivot_table
# → summary table
```

```python
# titanic example - survival
# what percentage of passenger survide by class and gender
pd.pivot_table(
    df,
    values="Survived",
    index="Pclass",
    columns="Sex",
    aggfunc="mean"
) * 100
```

```python
# multi index + titanic group by
result = df.groupby(
    ["Pclass", "Sex"]
)["Fare"].mean()
result #gives a multiindex series
```

```python
result.unstack()
# GroupBy → MultiIndex → Unstack
```

```python
# slicing multiindex
result = df.groupby(
    ["Pclass", "Sex"]
)["Fare"].mean()
result
```

```python
# to get class 1
result.loc[1]

# to get 1 class and male
result.loc[(1,"male")]
```

```python
# selecting multiple values
result.loc[
    [(1, "female"), (3, "male")]
]
```

```python
# xs - cross section
result.xs(1, level="Pclass")
```

```python
result.index.names
```

```python
# you can use the level name
result.xs("female", level="Sex")
# xs is a multiindex selection
```
