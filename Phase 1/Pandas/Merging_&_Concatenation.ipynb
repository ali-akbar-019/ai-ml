```python
# Horizontal Concatination
# axis = 1  - join columns
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed"]
})
df2 = pd.DataFrame({
    "Age": [21, 20, 22]
})

# ab karte ha column wise concat
result = pd.concat([df1, df2], axis=1)
print(result)
#  axis = 0 row wise hoti ha concatination and axis = 1 col wise kar deta ha
```

```python
# horizontal concatination with different indexes
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed"]
}, index=[0,1,2])
df1

df2 = pd.DataFrame({
    "Age" :[21,20,22]
}, index=[1,2,3])
pd.concat([df1,df2], axis=1)

# horizontal concat aligns data using the index
```

```python
# join = "inner"
pd.concat(
    [df1, df2],
    axis = 1,
    join="inner"
)
# only index taht exist in both
```

```python
# join = "outer"
pd.concat([df1, df2], axis=1, join="outer")
# keeps everything
# missing values become NaN
```

```python
# keys
# this is useful when you want to remeber where each dataframa came from
result = pd.concat(
    [df1, df2],
    keys = ["First", "Second"]
)
result
# this creates a multiindex
result.loc["First"]

```

```python
# basic merging
students = pd.DataFrame({
    "student_id": [1,2,3,4],
    "Name": ["Ali", "Sara", "Ahmed", "Zain"],
    "dept_id": [10,20,10,30]
})

departments = pd.DataFrame({
    "dept_id": [10,20,30],
    "Department": [
        "SE",
        "CS",
        "EE"
    ]
})


result = pd.merge(students, departments, on="dept_id")
result
```

```python
# on - tells pandas which colun should i use ot match teh dataframes
```

```python
# inner join
pd.merge(students, departments,
         on="dept_id", how="inner")
# this is the default join
# it keeps only the matching rows
```

```python
# left join
pd.merge(
    students,
    departments,
    on='dept_id',
    how="left"
)

# keep everything from the left dataframe
```

```python
# right join
pd.merge(students, departments, on="dept_id", how="right")
# keeos everything from the right table
```

```python
# outer join
pd.merge(
    students,
    departments,
    on="dept_id",
    how="outer"
)
# keep everything from both datafarmes
# left + right
# unmatched values become NaN
# join types
```

```python
# join types
# inner - matching rows only
# left - everything from left
# right  - everythign from right
# outer - everything from both

```

```python
# different column names
students = pd.DataFrame({
    "student_id": [1,2,3],
    "Name": ["Ali", "Sara", "Ahmed"],
    "department_id": [10,20,10]
})

departments = pd.DataFrame({
    "dept_id": [10, 20],
    "Department": ["SE", "CS"]
})

pd.merge(
    students,
    departments,
    left_on="department_id",
    right_on="dept_id"
)
# means k left wale table me department id available ha and ye right wale me dept id
```

```python
# multiple join columns
# someties one column isn't enough
sales = pd.DataFrame({
    "Store": ["A", "A", "B"],
    "Product":["Laptop", "Phone", "Laptop"],
    "Sales": [100,200,150]
})
prices = pd.DataFrame({
    "Store": ["A", "A", 'B'],
    "Product": ["Laptop", "Phone", "Laptop"],
    "Price": [1000,500, 900]
})

pd.merge(sales, prices, on="Store")
# we want store + product
pd.merge(sales, prices, on=["Store", "Product"])
```

```python
# duplicate column names
print(pd.merge(sales, prices , on="Store"))
print("\n"*3)
pd.merge(sales, prices, on="Store", suffixes=("_sales", "_prices"))

```

```python
# merge using index
pd.merge(
    df1,
    df2,
    left_index = True,
    right_index = True
)
# match the rows using their indexes
```

```python
# validate
# suppose you expect every student to mathc one department
pd.merge(
    students,
    departments,
    on="dept_id",
    validate="many_to_one"
)
```
