```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("train.csv")
df
```

```python
# the split apply combine concept
# data->split->apply->combine
df.groupby("Pclass")["Age"].mean()

df.head()
df.groupby("Ticket")["Name"].count()
```

```python
df.groupby("Sex")
# pandas only creates group
# to actually see result
# you must perform an operation
df.groupby("Sex")["Age"].mean()
```

```python
# sum
# total fare paid ny each class
df.groupby("Pclass")["Fare"].sum()
```

```python
# passenger in each class
df.groupby("Pclass")["PassengerId"].count()
```

```python
#
df.groupby("Pclass").size()
```

```python
# min
df.groupby("Pclass")["Age"].min()
```

```python
# max
df.groupby("Sex")["Age"].max()
```

```python
# median
df.groupby("Sex")["Fare"].median()
```

```python
# multiple aggregations agg()
df.groupby("Sex")["Age"].agg(
    ["min", "max", "mean", "median"]
)
```

```python
# different aggregation for different columns
df.groupby("Pclass").agg({
    "Age": "mean",
    "Fare": "sum"
})
```

```python
# multiple functions
df.groupby("Pclass").agg({
    "Age": ["min", "median", "sum"],
    "Fare": ["min", "max", "mean"]
})
```

```python
# grouping by multiple columns
df.groupby(
    ["Sex", "Pclass"]
)["Age"].mean()
# male in each p class
# female in each p class
```

```python
# get group
# we want only females
groups = df.groupby("Sex")
groups = groups.get_group("female")
groups.head()
```

```python
# iterating through groups
groups = df.groupby("Pclass")
for name, group in groups:
  print(name)
  print(group.head())
```

```python
# as index
df.groupby("Pclass")["Fare"].mean()

# remain a normal column
df.groupby("Pclass",
           as_index=False)["Fare"].mean()
  # pclass wont beome the index
```

```python
# apply custom func to all
df.groupby("Pclass")["Fare"].apply(lambda x: x.max() - x.min())
# group banao p class ki base pe
# mean k split kar do pclass pe
# then select karo fare wala col
# then fare k har col se max - min nikalo
# end pe har aik pclass k leye fare ki value de do
```

```python
df.groupby("Pclass")["Age"].apply(
    lambda x: x.isnull().sum()
)
# it counts missing ages in each class
```

```python
# transform
df["ClassAvgAge"] = df.groupby("Pclass")["Age"].transform("mean")
df
```
