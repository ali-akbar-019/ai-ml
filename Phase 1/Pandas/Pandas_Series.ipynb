```python
# Question 1 – Student Marks Analysis
m = np.random.randint(40,100, 20)
students_name = ["ali", "zara", "ahmed", "fatima", "hassan", "ayesha", "umar", "sana", "hamza", "maryam", "bilal", "hira", "usman", "mehak", "rizwan", "naila", "asif", "sidra", "sajid", "kiran"]
s = pd.Series(m, index=students_name)
print("--Head:")
print(s.head())
print("--Tail:")
print(s.tail())
print("--Random:")
print(s.sample(4))
print("-- Sort by name")
print(s.sort_index())
print("-- Describe")
print(s.describe())
print('-- Student  who scored greater than 85')
print(s[s >= 85])
print("-- Student who scored between 70 and 90")
print(s[(s >= 70) & (s <= 90)])
print("-- Fancy indexing")
print(s[[1,2,3]])
```
